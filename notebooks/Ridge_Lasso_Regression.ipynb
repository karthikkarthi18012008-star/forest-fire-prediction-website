```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('DATA/Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
df.head()
```

```python
##data cleaning
df.isnull().sum()
df[df.isnull().any(axis=1)]
```

```python
##dividing the data into two sets
df = df.reset_index(drop=True)

df.loc[:122, "Region"] = 0
df.loc[123:, "Region"] = 1
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
##removing the nulls
df = df.dropna().reset_index(drop=True)
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df = df.drop(122).reset_index(drop=True)
df.iloc[[122]]
```

```python
##fix spaces in columns names
df.columns
```

```python
##fix spaces in columns names
df.columns=df.columns.str.strip()
df.columns
```

```python
##change the data types to int for the required columns
df.head()
df[['day','month','year','Temperature','RH','Ws']] = df[['day','month','year','Temperature','RH','Ws']].astype(int)
```

```python
##change the other columns to float
# Convert all columns except Classes to numeric
for col in df.columns:
    if col != 'Classes':
        df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.info()
```

```python
df.describe()
```

```python
# EDA & Feature Engineering
df.to_csv("DATA/Algerian_forest_fires_cleaned_dataset.csv",index=False)
```

```python
##exploratory data analysis
df_copy = df
df_copy = df.drop(['day','month','year'],axis=1)
df_copy.head()
```

```python
##encode the categories in classes
df_copy['Classes'] = np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy['Classes'].value_counts()
```

```python
##plot density plot for all features
df_copy.hist(bins=50,figsize=(10,15))
```

```python
## Percentage for Pie Chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
# plotting piechart
classlabels=["Fire","Not Fire"]
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.title("Pie Chart of Classes")
plt.show()
```

```python
##correlation
df_copy.corr()
```

```python
##plotting heatmap
sns.heatmap(df_copy.corr(),annot=True)
```
